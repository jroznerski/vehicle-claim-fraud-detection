```python
import pandas as pd

df = pd.DataFrame({
    'date': ['2024-01-05', '2024-01-20', '2024-02-10', '2024-02-25',
             '2024-03-05', '2024-03-20', '2024-01-10', '2024-02-15',
             '2024-03-10', '2024-03-25'],
    'product_id': [1, 1, 1, 1, 1, 1, 2, 2, 2, 2],
    'category': ['electronics', 'electronics', 'electronics', 'electronics',
                 'electronics', 'electronics', 'furniture', 'furniture',
                 'furniture', 'furniture'],
    'revenue': [5000, 6000, 4500, 7000, 5500, 6500, 2000, 2500, 1800, 3000],
    'units': [2, 3, 2, 4, 3, 3, 5, 6, 4, 7]
})
```

```python
df.head()
```

```python
df['date'] = pd.to_datetime(df['date'])
df['month'] = df['date'].dt.to_period('M')
df['avg_price'] = df['revenue'] / df['units']
df.head()
```

```python
df_grouped = df.groupby(['product_id', 'month']).agg({
    'revenue': 'sum',
    'units': 'sum',
    'avg_price': 'mean'
})
df_grouped
```

```python
df = df.sort_values(by=['product_id', 'date'])
df['revenue_prev'] = df.groupby('product_id')['revenue'].shift(1)
df['revenue_growth'] = df.groupby('product_id')['revenue'].pct_change() * 100
df.head()
```

```python
df_category_sum_revenue = df.groupby('category')['revenue'].sum()
final = df_category_sum_revenue.idxmax()
final
```

```python
df_average_month = df.groupby('month')['avg_price'].mean().reset_index()
df_average_month_max = df_average_month[df_average_month['avg_price'] == df_average_month['avg_price'].max()]
df_average_month_max
```
