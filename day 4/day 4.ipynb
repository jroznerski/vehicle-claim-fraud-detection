```python
import pandas as pd

df = pd.DataFrame({
    'date': ['2024-01-10', '2024-01-15', '2024-02-05', '2024-02-20',
             '2024-03-01', '2024-03-15', '2024-03-28'],
    'product': ['laptop', 'phone', 'laptop', 'tablet',
                'phone', 'laptop', 'tablet'],
    'sales': [3500, 2000, 3500, 1500, 2000, 3500, 1500],
    'units': [2, 5, 1, 3, 4, 2, 6]
})
```

```python
df.head()
```

```python
df['date'] = pd.to_datetime(df['date'])
df['month'] = df['date'].dt.month
df['revenue'] = df['sales'] * df['units']
print(df)
```

```python
df_grouped = df.groupby(['month', 'product']).agg({'revenue': 'sum', 'units': 'sum'}).reset_index()
print(df_grouped)
```

```python
product_revenue = df.groupby('product')['revenue'].sum()
best_product = product_revenue.idxmax()
print(best_product)
```

```python
import pandas as pd

df_sales = pd.DataFrame({
    'seller_id': [1, 1, 1, 2, 2, 2, 3, 3, 3],
    'month': ['2024-01', '2024-02', '2024-03',
              '2024-01', '2024-02', '2024-03',
              '2024-01', '2024-02', '2024-03'],
    'revenue': [10000, 12000, 9000, 15000, 14000, 16000, 8000, 9000, 11000]
})

df_sellers = pd.DataFrame({
    'seller_id': [1, 2, 3],
    'name': ['Anna', 'Bartek', 'Celina'],
    'region': ['North', 'South', 'North']
})
```

```python
df_merged = pd.merge(df_sales, df_sellers, on='seller_id', how='left')
print(df_merged)
```

```python
df_merged = df_merged.sort_values(by=['seller_id', 'month'])
df_merged['revenue_prev'] = df_merged.groupby('seller_id')['revenue'].shift(1)
df_merged['growth'] = df_merged.groupby('seller_id')['revenue'].pct_change() * 100
print(df_merged)
```

```python
df_avg_region = df_merged.groupby('region')['revenue'].mean().reset_index()
print(df_avg_region)
```

```python
#need to find a months when every seller had growth > 0
df_filtered = df_merged[df_merged['growth'] > 0]
counts = df_filtered.groupby('month')['seller_id'].count()
counts[counts==3]
```

```python
print(df_merged[['name', 'month', 'revenue', 'revenue_prev', 'growth']])
```

```python
import pandas as pd

df_orders = pd.DataFrame({
    'order_id': range(1, 10),
    'customer_id': [1, 1, 1, 2, 2, 2, 3, 3, 3],
    'month': ['2024-01', '2024-02', '2024-03',
              '2024-01', '2024-02', '2024-03',
              '2024-01', '2024-02', '2024-03'],
    'amount': [500, 700, 600, 1000, 900, 1100, 300, 350, 400]
})

df_customers = pd.DataFrame({
    'customer_id': [1, 2, 3],
    'name': ['Anna', 'Bartek', 'Celina'],
    'segment': ['premium', 'premium', 'standard']
})
```

```python
df_merged_orders = pd.merge(df_orders, df_customers, on='customer_id', how='left')
df_merged_orders.head()
```

```python
df_merged_orders['amount_prev'] = df_merged_orders.groupby('customer_id')['amount'].shift(1)
df_merged_orders['growth'] = df_merged_orders.groupby('customer_id')['amount'].pct_change() * 100
df_merged_orders.head()
```

```python
df_merged_orders_avg_segment = df_merged_orders.groupby('segment')['amount'].mean().reset_index()
df_merged_orders_avg_segment
```

```python
df_merged_growth_filtered = df_merged_orders[df_merged_orders['growth'] > 0]
counts = df_merged_growth_filtered.groupby('customer_id')['month'].count()
n_months = 2 
winning_customers = counts[counts == n_months]
winning_customers
```
