```python
import pandas as pd

df = pd.DataFrame({
    'name': ['Anna', 'Bartek', 'Celina', 'Darek', 'Ewa'],
    'department': ['IT', 'HR', 'IT', 'HR', 'IT'],
    'salary': [5000, 3000, 7000, 4500, 6000],
    'years_experience': [3, 5, 8, 2, 6]
})
```

```python
df.head()
```

```python
df_filtered = df[df['department'] == 'IT']
df_filtered.head()
```

```python
avg_salary = df_filtered['salary'].mean()
print(avg_salary)
```

```python
avg_ye_exp = df_filtered['years_experience'].mean()
print(avg_ye_exp)
```

```python
df_filtered[['salary','years_experience']].mean()
```

```python
df.groupby('department').agg(
    srednia = ('salary','mean'),
    max_experience = ('years_experience','max'),
    number_of_emplyess = ('name', 'count')
)
```

```python
df_sales = pd.DataFrame({
    'month': ['2024-01', '2024-02', '2024-03', '2024-04', '2024-05'],
    'sales': [100, 120, 90, 150, 130]
})

df_sales.head()
```

```python
df_sales['sales_prev'] = df_sales['sales'].shift(1).fillna('Brak')
df_sales['sales_diff'] = df_sales['sales'] - df_sales['sales'].shift(1)

df_sales.head()
```
