```python
def calculate_bonus(salary, performance_rating):
    if performance_rating not in (1,2,3,4,5):
        raise ValueError(f"performance_rating musi być od 1 do 5, dostałem: {performance_rating}")
    if performance_rating == 5:
        bonus =  0.2*salary
    elif performance_rating == 4:
        bonus =  0.1 * salary
    else:
        bonus = 0
    return bonus
```

```python
calculate_bonus(5000, 5)
```

```python
calculate_bonus(5000, 3)
```

```python
calculate_bonus(5000, 8)
```

```python
salaries = [3000, 5000, 7000, 2000, 8000, 4500]
new_salaries = []

for salary in salaries:
    if salary > 4000:
        salary = salary + 0.1*salary
        new_salaries.append(salary)

new_salaries
```

```python
employees = [
    {'name': 'Anna', 'salary': 5000, 'department': 'IT'},
    {'name': 'Bartek', 'salary': 3000, 'department': 'HR'},
    {'name': 'Celina', 'salary': 7000, 'department': 'IT'},
    {'name': 'Darek', 'salary': 4500, 'department': 'HR'},
]
```

```python
sorted_employees = sorted(employees, key=lambda x: x['salary'], reverse=True)
sorted_employees
```

```python
filtered_employees = list(filter(lambda x: x['department']=='IT', employees))
filtered_employees
```
