```python
transactions = [
    {'id': 1, 'amount': 250, 'status': 'completed'},
    {'id': 2, 'amount': 1500, 'status': 'completed'},
    {'id': 3, 'amount': 80, 'status': 'failed'},
    {'id': 4, 'amount': 3000, 'status': 'completed'},
    {'id': 5, 'amount': 500, 'status': 'failed'},
]
```

```python
def get_high_value_completed(x, threshold):
    a = [a['id'] for a in x if a['status'] == 'completed' and a['amount'] > threshold]
    return a
```

```python
get_high_value_completed(transactions, 200)
```

```python
import pandas as pd

df = pd.DataFrame({
    'seller': ['Anna', 'Anna', 'Anna', 'Bartek', 'Bartek', 'Celina'],
    'product': ['A', 'B', 'A', 'B', 'A', 'A'],
    'sales': [100, 200, 150, 300, 120, 400],
    'returned': [0, 1, 0, 0, 1, 0]
})

df.head()
```

```python
df.groupby('seller').agg(
    sum_sales=('sales', 'sum'),
    sum_returned=('returned', 'sum'),
    prc_returned=('returned', lambda x: x.sum() / x.count())
)
```

```python
df_orders = pd.DataFrame({
    'order_id': [1, 2, 3, 4, 5],
    'customer_id': [101, 102, 101, 103, 102],
    'amount': [250, 180, 320, 90, 410]
})

df_customers = pd.DataFrame({
    'customer_id': [101, 102, 104],
    'name': ['Anna', 'Bartek', 'Celina'],
    'city': ['Warsaw', 'Krakow', 'Gdansk']
})
```

```python
merged_1 = pd.merge(df_orders, df_customers, on='customer_id', how='inner')

merged_1.head()
```

```python
merged_2 = pd.merge(df_orders, df_customers, on='customer_id', how='left')

merged_2.head()
```

```python
df_sales3 = pd.DataFrame({
    'seller': ['Anna', 'Anna', 'Anna', 'Bartek', 'Bartek', 'Bartek'],
    'month': ['2024-01', '2024-02', '2024-03', '2024-01', '2024-02', '2024-03'],
    'sales': [100, 120, 90, 200, 180, 220]
})

df_sales3.head()
```

```python
df_sales3['sales_prev'] = df_sales3.groupby('seller')['sales'].shift(1)
df_sales3['sales_diff'] = df_sales3['sales'] - df_sales3['sales_prev']
df_sales3['is_better'] = df_sales3['sales_diff'] > 0

df_sales3.head()
```
