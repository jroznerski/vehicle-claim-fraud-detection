```python
class BankAccount():
    def __init__(self, owner, balance=0):
        if not isinstance(owner, str):
            raise ValueError("błędna nazwa użytkownika")
        self.owner = owner
        self.balance = balance
    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
        else: raise ValueError
    def withdraw(self,amount):
        if amount <= 0 or amount > self.balance:
            raise ValueError("wartość ma być powyżej 0")
        else: self.balance -= amount
    def __str__(self):
        return f"Konto {self.owner}: {self.balance:.2f} PLN"
```

```python
class SavingsAccount(BankAccount):
    def __init__(self, owner, balance, interest_rate):
        if not isinstance(interest_rate, (int,float)) or interest_rate < 0:
            raise ValueError("bład danych")
        super().__init__(owner, balance)
        self.interest_rate = interest_rate
    def apply_interest(self):
        self.balance = self.balance + self.balance * self.interest_rate
    def __str__(self):
        return f"Konto oszczędnościowe {self.owner} ({self.interest_rate * 100} %): {self.balance} PLN"
```

```python
acc = SavingsAccount("Anna", 1000, 0.05)
acc.deposit(500)
acc.apply_interest()
print(acc)  # "Konto oszczędnościowe Anna (5.0%): 1575.00 PLN"

#acc.withdraw(10000)  # powinno rzucić ValueError
```

```python
Napisz klasę BankAccount która:

W __init__ przyjmuje owner (string) i balance (float, domyślnie 0)
Ma metodę deposit(amount) — dodaje środki, ale rzuca ValueError jeśli amount <= 0
Ma metodę withdraw(amount) — odejmuje środki, ale rzuca ValueError jeśli amount > balance albo amount <= 0
Ma metodę __str__ — zwraca "Konto Anna: 1500.00 PLN"

Potem napisz klasę SavingsAccount która dziedziczy po BankAccount i:

W __init__ przyjmuje dodatkowo interest_rate (np. 0.05 = 5%)
Ma metodę apply_interest() — zwiększa balance o balance * interest_rate
Nadpisuje __str__ — zwraca "Konto oszczędnościowe Anna (5.0%): 1500.00 PLN"
```
