```python
def calculate_bonus(salary, performance_rating):
    if performance_rating not in (1,2,3,4,5):
        raise ValueError("rating musi być od 1 do 5")
    if type(salary) is not int:
        raise TypeError ("salary musi być liczbą")
    if performance_rating == 5:
        bonus =  0.2*salary
    elif performance_rating == 4:
        bonus =  0.1 * salary
    else:
        bonus = 0
    return bonus
```

```python
try:
    calculate_bonus("pięć tysięcy", 3)
except TypeError as e:
    print(f"Błąd typu: {e}")
except ValueError as e:
    print(f"Błąd wartości: {e}")
```

```python
def even_numbers(n):
    for i in range(0, n+1):
        if i%2 == 0:
            yield i
```

```python
for num in even_numbers(10):
    print(num)  # 0, 2, 4, 6, 8, 10
```
