```python
import pandas as pd

df = pd.DataFrame({
    'employee_id': [1, 1, 1, 2, 2, 2, 3, 3, 3],
    'month': ['2024-01', '2024-02', '2024-03',
              '2024-01', '2024-02', '2024-03',
              '2024-01', '2024-02', '2024-03'],
    'sales': [100, 150, 120, 200, 180, 220, 90, 95, 110],
    'target': [120, 120, 120, 200, 200, 200, 100, 100, 100]
})

df.head()
```

```python
df['target_hit'] = df['sales'] > df['target']

df.head()
```

```python
df = df.sort_values(['employee_id','month'])

df['sales_prev'] = df.groupby('employee_id')['sales'].shift(1)

df.head()
```

```python
df['sales_growth'] = (df['sales'] / df['sales_prev']) * 100

df.head()
```

```python
df.groupby('employee_id')['target_hit'].agg('sum')
```

```python
df.groupby('employee_id').agg(
    most_sales = ('sales', 'mean')
).idxmax()
```
